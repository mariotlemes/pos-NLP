```python
# Receber uma lista de notas de alunos (como uma tupla). Determine um conjunto de no mínimo 8 notas
tuplas_notas = (1.0, 6.0, 6.5, 8.8, 9.0, 6.3, 5.5, 10.0) # 8 notas
# print(tuplas_notas)

import numpy as np

# Converta essa tupla em um array NumPy
tuplas_notas_array = np.array(tuplas_notas)
# print(tuplas_notas_array)

# Calcule a media das notas
print(f"A media das notas é: {tuplas_notas_array.mean()}.")

# Encontre a maior e a menor nota
print(f"A maior nota é: {tuplas_notas_array.max()}.")
print(f"A menor nota é: {tuplas_notas_array.min()}.")

# Determine quantos alunos tiraram nota acima da média
mask = tuplas_notas_array > tuplas_notas_array.mean()
# print(mask)
notas_maiores = tuplas_notas_array[mask]
print(f"As notas maiores que a media são: {notas_maiores}.")

# Aplicar um acréscimo de 10% a todas as notas usando np.multiply() e np.round() (sem ultrapassar o limite de 10,0)
a = 1.1 # valor para multiplicar
tuplas_notas_array_com_aumento = np.multiply(tuplas_notas_array.round(2), a) # valores corrigidos em 10%
tuplas_notas_array_tratadas = np.where(tuplas_notas_array_com_aumento > 10, 10, tuplas_notas_array) # tratamento para não ultrapassar 10,0
print(f"Notas com acréscimo de 10%: {tuplas_notas_array_tratadas}.")

# Usando a biblioteca Matplotlib, gere um gráfico de barras que mostre: As notas originais; As notas ajustadas (com acréscimo de 10%). Uma linha horizontal representando a média.

import matplotlib.pyplot as plt

# Dados
x =  tuplas_notas_array
y = tuplas_notas_array_com_aumento

# plt.figure(figsize=(6,6))
plt.plot(x, y, label= 'Relação entre notas originais e ajustadas')
plt.legend() # mostrar a legenda
plt.xlabel("Notas originais")
plt.ylabel("Notas reajustadas (10%)")

# Linha horizontal com a média
plt.axhline(y=tuplas_notas_array.mean().round(2), color='r', linestyle='-', label='Media')
plt.legend()

plt.show()
```
