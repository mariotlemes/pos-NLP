```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

url = 'https://raw.githubusercontent.com/vodena/GRIDouble/refs/heads/main/data/Data_Cacak.csv'
energia_solar = pd.read_csv(url)
energia_solar.head()
```

```python
# quantidade de instâncias
print(f"Instâncias: {energia_solar.shape[0]}")
print(f"Atributos: {energia_solar.shape[1]}")
```

```python
# verificar valores nulos
energia_solar.isnull().sum()
```

```python
energia_solar[energia_solar.duplicated()]
```

```python
energia_solar.info()
```

```python
# plt.suptitle("Gráfico de calor das variáveis numéricas")
dados_numericos = energia_solar.select_dtypes(include=['int', 'float', 'bool'])

fig, ax = plt.subplots(figsize=(10, 8))  # defina o tamanho
ax = sns.heatmap(dados_numericos.corr(), annot=True, cmap="Greens", annot_kws={"size": 15, "fontweight": "bold"})

fig.tight_layout()
fig.savefig("teste.pdf", bbox_inches="tight")
```
