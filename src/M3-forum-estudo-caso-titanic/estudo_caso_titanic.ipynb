```python
# Crie uma nova coluna chamada "family_size", somando as colunas SibSp (número de irmãos/cônjuges a bordo)
# e Parch (número de pais/filhos a bordo). Depois, utilize um gráfico de violino para ver se o tamanho da família influenciou a taxa
# de sobrevivência.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# carregamento da base de dados
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv'
titanic = pd.read_csv(url)

# adicionar nova coluna com o tamanho da família - resultante das colunas 'sibsp' e 'parch'
titanic['family_size'] = titanic['sibsp'] + titanic['parch']

# Gráfico de violino para ver se o tamanho da família influenciou a taxa de sobrevivência.
sns.violinplot(x='survived', y='family_size', data=titanic)
plt.title('Relação entre tamanho da família e sobrevivência')
plt.xlabel('Sobreviveu')
plt.ylabel('Tamanho da família')
plt.show()

print("Através do gráfico de violino, percebe-se que poucas famílias entre 4 e 7 integrantes sobreviveram.")
print("Famílias com mais de 7 integrantes não sobreviveram")
print("A maioria das famílias com 1, 2 ou 3 integrantes sobreviveram")
print("A porcentagem de pessoas sozinhas não sobreviventes é maior que pessoas sozinhas sobrevientes")

# Crie um gráfico de dispersão (scatterplot) para visualizar a relação entre idade e tarifa paga,
# colorindo os pontos de acordo com a sobrevivência. Passageiros mais jovens pagaram tarifas mais altas?
# Relação entre idade, tarifa e sobrevivência

plt.scatter(titanic['age'], titanic['fare'], alpha=0.5,
            # s=titanic['pclass']*20,  # Tamanho do ponto baseado em pclass
            c=titanic['survived'], cmap='coolwarm')  # Cor baseada na sobrevivência
plt.colorbar(label='Sobreviveu')
plt.title('Relação entre Idade, Tarifa e Sobrevivência')
plt.xlabel('Idade')
plt.ylabel('Tarifa')
plt.show()

print("Passageiros mais jovens (entre 0 e 8 anos) que sobreviveram pagaram tarifas abaixo de $50")
print("Uma grande quantidade de passageiros (entre 20 e 50) anos que pagaram tarifas abaixo de $50 não sobreviveram ")
```
