```python
# Crie uma nova coluna chamada "family_size", somando as colunas SibSp (número de irmãos/cônjuges a bordo)
# e Parch (número de pais/filhos a bordo). Depois, utilize um gráfico de violino para ver se o tamanho da família influenciou a taxa
# de sobrevivência.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# carregamento da base de dados
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv'
titanic = pd.read_csv(url)
# titanic.head()

# quantidade de instâncias de pessoas com irmãos/cônjugues a bordo com shape[0]
tamanho_coluna_sibsp = titanic.loc[titanic['sibsp'] == 1].shape[0]
print(f"Quantidade de pessoas com irmãos/cônjuges a bordo: {tamanho_coluna_sibsp}")

# quantidade de instâncias de pessoas com pais/filhos a bordo\
tamanho_coluna_parch = titanic.loc[titanic['parch'] == 1].shape[0]
print(f"Quantidade de pessoas com pais/filhos a bordo: {tamanho_coluna_parch}")

# tamanho da família
family_size = tamanho_coluna_sibsp + tamanho_coluna_parch
print(f"Tamanho da família: {family_size}")

# adicionar a coluna family_size no dataset titanic


# Gráfico de violino para ver se o tamanho da família influenciou a taxa de sobrevivência
```
