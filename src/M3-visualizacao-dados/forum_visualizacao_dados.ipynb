```python
 # 1. Apresente a distribuição do comprimento da pétala (petal_length) para cada espécie usando um gráfico de violino.

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/uiuc-cse/data-fa14/gh-pages/data/iris.csv" # link do csv no github
iris = pd.read_csv(url) #lendo o csv e colocando no Dataframe chamando iris

plt.figure(figsize=(10,6)) #definir o tamanho do plot

sns.violinplot(x='species', #eixo x
               y='petal_length', #eixo y
               data=iris,#carregamento dos dados
               palette='Pastel1', #cores dos violinos
               hue='species', #as cores dos violinos são determinados pelas espécies
               legend=False) #definir como False pois já aparece a legenda no eixo x)

plt.title('Representação gráfica do comprimento das pétalas segundo espécie')
plt.xlabel("Espécies")
plt.ylabel("Comprimento das pétalas (cm)")

plt.show()

# 2. Crie um gráfico de barras mostrando a média do comprimento da pétala (petal_length) para cada espécie.
# Qual espécie tem o maior valor médio? Resposta: virginica

plt.figure(figsize=(10,6)) #definir o tamanho do plot

sns.barplot(
    x='species',
    y='petal_length',
    data=iris,
    palette='Pastel1',
    hue='species',
    legend=False)

plt.title("Representação gráfica da média do comprimento das pétalas segundo espécie.")
plt.xlabel("Espécies")
plt.ylabel("Comprimento das pétalas (cm)")
plt.show()

print("A espécie com maior valor médio de comprimento de pétala é a virginica")

# 3. Utilize um boxplot para identificar possíveis outliers no comprimento da pétala (petal_length).
# Quais espécies possuem mais outliers? Esses valores podem indicar alguma característica biológica específica?

plt.figure(figsize=(10,6)) #definir o tamanho do plot

sns.boxplot(x='species',
            y='petal_length',
            data=iris,
            palette='Pastel1',
            hue='species',
            legend=False)

plt.title("Representação gráfica do comprimento das pétalas por meio de um boxplot.")
plt.xlabel("Espécies")
plt.ylabel("Comprimento das pétalas (cm)")
plt.show()

print("Note que as bolinhas no gráfico de box plot representam os outliers. Portanto, a espécie com maior número de outliers é a setosa (total de 3). A espécie versicolor apresenta somente 1 incidência de outlier. Acredito que pode ser um fator relacionado a genética que define essa anomalia.")
```
