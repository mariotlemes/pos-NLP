#### NER (Named Entitity Recognition) - Reconhecimento de Entidades Nomeadas

Colab de referência: https://colab.research.google.com/drive/1Tn8zErEyoNrrNmqJMijQTM-IcjswXE2y#scrollTo=bB7txcOYkYN8

NER é a técnica de identificar e classificar automaticamente, em um texto, quais partes correspondem a entidades nomeadas (pessoas, empresas, locais, datas, quantias, etc.).

```python
import spacy
from spacy.cli import download

# Garantir que o modelo pt_core_news_sm está instalado
try:
    nlp = spacy.load('pt_core_news_sm')
except OSError:
    download('pt_core_news_sm')
    nlp = spacy.load('pt_core_news_sm')

# Texto de exemplo em português
text = "A Apple anunciou a abertura de uma nova loja em Brasília, capital do Brasil."

# Processar o texto
doc = nlp(text)

# Imprimir entidades reconhecidas
for ent in doc.ents:
    print(ent.text, ent.label_)
```
