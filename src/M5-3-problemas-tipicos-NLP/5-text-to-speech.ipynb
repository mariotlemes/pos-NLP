```python
import pyttsx3
from deep_translator import GoogleTranslator

# Lista de frases em português
texts = [
    "O gato está no tapete.",
    "A casa é grande e bonita.",
    "Ele gosta de jogar futebol.",
    "Vamos ao cinema hoje à noite.",
    "Ela está estudando para o exame.",
    "O gato preto da Renata dorme neste momento."
]

# Função para traduzir texto usando DeepTranslator
def translate_text(text, lang='en'):
    return GoogleTranslator(source='pt', target=lang).translate(text)

# Função para converter texto em áudio e reproduzir
def text_to_speech(text, lang='en'):
    engine = pyttsx3.init()
    engine.setProperty('rate', 150)  # Velocidade da fala
    engine.setProperty('volume', 1)  # Volume (0.0 a 1.0)
    engine.setProperty('voice', lang)  # Idioma (use "english" para inglês ou "brazil" para português)
    engine.say(text)
    engine.runAndWait()

# Traduzir para inglês e francês e gerar áudio
for text in texts:
    # Tradução para o inglês
    translation_en = translate_text(text, 'en')
    print(f'Texto original: {text}')
    print(f'Tradução para o inglês: {translation_en}')
    text_to_speech(translation_en, 'english')  # Pronúncia em inglês
    print()

    # # Tradução para o francês
    # translation_fr = translate_text(text, 'fr')
    # print(f'Texto original: {text}')
    # print(f'Tradução para o francês: {translation_fr}')
    # text_to_speech(translation_fr, 'french')  # Pronúncia em francês
    # print()
```
