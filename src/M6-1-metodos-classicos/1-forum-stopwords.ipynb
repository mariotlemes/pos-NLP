```python
# Testes preliminares

import pandas as pd
df = pd.DataFrame({
    "texto": ["Olá mundo", "Este é um exemplo", "Contando palavras aqui"]
})

df["contagem_palavras"] = df["texto"].str.split().str.len()
print(df)
```

```python
import pandas as pd

olist = pd.read_csv('olist.csv')

# olist

freq_palavras = pd.Series(
    " ".join(olist.loc[:, "review_text"].astype(str)).lower().split()
).value_counts()

print(freq_palavras)
# olist
```

```python
# Coluna antes de ser modificada
olist['review_text']
```

```python
stopwords_personalizadas = ["o", "e", "produto", "a", "de"]

olist.loc[:, 'review_preproc'] = olist['review_text'].apply(lambda x: ' '.join([termo for termo in x.split() if termo not in stopwords_personalizadas]))

olist['review_preproc']
```
