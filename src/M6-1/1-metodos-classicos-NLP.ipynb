```python
# Filtrar o DataFrame para colunas selecionadas
df = df_original[['review_text', 'review_text_processed', 'polarity']]
df
```

```python
#### Normalizar capitalização
exemplo = 'EXCELENTE!! Valeu demais passar um tempo pesquisando preços, pois encontrei esse ótimo carregador de celular, nota 10.'
print ("Antes de normalizar: ")
print(exemplo)
exemplo = exemplo.lower()
print ("Depois de de normalizar: \n")
print(exemplo)
```

#### Normalizando a coluna review_text inteira

```python
# df['review_preproc'] = df['review_text'].apply(lambda x: x.lower())
df.loc[:, 'review_preproc'] = df['review_text'].apply(lambda x: x.lower())
df
```

#### Retirar pontuação

```python
# Opção 1
exemplo_sem_punct_opcao1 = "".join(u for u in exemplo if u not in ("?", ".", ";", ":", "!", ","))
print(exemplo_sem_punct_opcao1)
```

```python
# Opcão 2
from nltk.tokenize import RegexpTokenizer

# com '\w+' selecionamos apenas tokens de palavras ou dígitos
tokenizer = RegexpTokenizer(r'\w+')

exemplo_sem_punct_opcao2 = ' '.join(tokenizer.tokenize(exemplo))
print(exemplo_sem_punct_opcao2)
```

#### Retirando a pontuação da coluna inteira

```python
# df['review_preproc'] = df['review_preproc'].apply(lambda x: ' '.join(tokenizer.tokenize(x))) # opção do colab
df.loc[:, 'review_preproc'] = df['review_preproc'].apply(lambda x: ' '.join(tokenizer.tokenize(x))) # opção alternativa, garante que df nãoé um subconjunto e evita warning
print(df)
```

#### Retirar acentos

```python
# Opção 1
repl = str.maketrans(
    "áéúíó",
    "aeuio"
)
palavra = 'ótimo'
palavra.translate(repl)
```

```python
# Opção 2: Usar lib unidecode, https://github.com/avian2/unidecode.
try:
    import unidecode
    import logging
    libraries_installed = True
except ImportError:
    libraries_installed = False

if not libraries_installed:
    !pip install unidecode

try:
    import unidecode
    print(f"✅ Biblioteca está instalada e foi importada com sucesso!\n")
except ImportError as e:
    print(f"❌ Erro ao instalar a biblioteca: {e}\n")

exemplo_sem_acento = ' '.join([unidecode.unidecode(termo) for termo in exemplo_sem_punct_opcao2.split()])
exemplo_sem_acento
```

#### Retirando os acentos da coluna inteira

```python
df.loc[:, 'review_preproc'] = df['review_preproc'].apply(lambda x: ' '.join([unidecode.unidecode(termo) for termo in x.split()]))
df
```

#### Retirar stopwords

```python
# Opção 1: criar uma lista manual
stopwords = ['de', 'para', 'uma', 'o', 'e'] # aqui listamos as palavras que mais se repetem, de acordo com o domínio de aplicação
```

```python
# Opção 2: usar uma lista de stopwords compilada pela biblioteca NLTK
import nltk
nltk.download('stopwords')
stopwords = nltk.corpus.stopwords.words('portuguese')
# stopwords

print(f"Com stopwords: {exemplo_sem_acento}\n")
exemplo_sem_stopwords = ' '.join([termo for termo in exemplo_sem_acento.split() if termo not in stopwords])
print(f"Sem stopwords: {exemplo_sem_stopwords}")
```

#### Remover stopwords da coluna review_preproc por inteiro

```python
df.loc[:, 'review_preproc'] = df['review_preproc'].apply(lambda x: ' '.join([termo for termo in x.split() if termo not in stopwords]))
df
```

#### Stemmização da coluna inteira

```python
nltk.download('rslp')
stemmer = nltk.stem.RSLPStemmer()
```

```python
df.loc[:, 'review_preproc'] = df['review_preproc'].apply(lambda x: ' '.join(stemmer.stem(termo) for termo in x.split()))
df
```
